```python
# correct column data types
df['Date of Purchase'] = pd.to_datetime(df['Date of Purchase'])
df['First Response Time'] = pd.to_datetime(df['First Response Time'])
```

## Dealing with missing values

We see same number of null values in Customer Satisfaction Rating, Resolution and Time to Resolution
so we analyse these first.

```python
df['Ticket Status'].value_counts()
```

- Sum of unresolved tickets (open + pending customer response) is 5700.
- And we have same amount of missing values in Customer Satisfaction Rating,
- Resolution and Time to Resolution.
- So, we check if the missing values are in unresolved tickets only and then
- move to analyse missing values in 'Time to resolution'.

```python
null_rows = df[
    df['Resolution'].isnull() &
    df['Time to Resolution'].isnull() &
    df['Customer Satisfaction Rating'].isnull()
]

print(null_rows[['Ticket Status']].value_counts())
```

```python
null_rows[['Ticket Status', 'Resolution', 'Customer Satisfaction Rating']]
```

```python
# check if any closed ticket has any null value in any of these three column
closed_with_nulls = df[
    (df['Ticket Status'] == 'Closed') &
    (
        df['Resolution'].isnull() |
        df['Time to Resolution'].isnull() |
        df['Customer Satisfaction Rating'].isnull()
    )
]

print(closed_with_nulls.shape)
closed_with_nulls.head()
```

This shows we have no null values in Customer Satisfaction Rating, Resolution and Time to Resolution
where ticket has been resolved (closed).

Let's now analyse First Response Time.

```python
df['First Response Time'].isnull().sum()
```

```python
null_response_df = df[df['First Response Time'].isnull()]
print(null_response_df[['Ticket Status']].value_counts())
```

- This shows we have null values in First Response Time where ticket is open.
- Conclusively, these are new tickets which are not yet alined in resolution/ not picked up.

## Data Cleaning Summary

The missing values in the dataset were found to be structurally meaningful rather than random.  
Null values in `Resolution`, `Time to Resolution`, and `Customer Satisfaction Rating` occurred only for unresolved tickets (`Open` and `Pending Customer Response`) and in `First Response Time` null values occurred in `open tickets`, indicating business-process-driven missingness.
 

Therefore, no direct imputation or removal was performed during the cleaning phase. Apart from analysing and validating missing values, the dataset required minimal cleaning.
